```python
import pandas as pd
```

```python
df=pd.read_csv('clean_vehicle_data.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.drop('InteriorColour',axis=True,inplace=True)
```

```python
numerical_value = [
    'EngineSizeL',
    'Cylinders',
    'FuelConsumption_L100km',
    'Doors',
    'Seats'
]
for i in numerical_value:
    df[i]=df[i].fillna(df[i].median())
```

```python
df.isna().sum()
```

```python
catagorical_value=[
    'Transmission',
    'FuelType',
    'ExteriorColour'
]
for i in catagorical_value:
    df[i]=df[i].fillna(df[i].mode()[0])
```

```python
df.isna().sum()
```

```python
df['Doors']=df['Doors'].astype(int)
```

```python
df['Seats']=df['Seats'].astype(int)
```

```python
df.info()
```

```python
a=['Brand','Model','UsedOrNew','Transmission','DriveType','FuelType','ExteriorColour','State','BodyType']
df[a]=df[a].apply(lambda x: x.str.strip())
```

```python
df['UsedOrNew'].unique()
```

```python
df['UsedOrNew'].value_counts()
```

```python
df = df.rename(columns={'UsedOrNew' : 'Car_Condition'})
```

```python
df.info()
```

```python
df['Car_Condition'].value_counts()
```

```python
co=df['Doors'].value_counts()
co
# 
```

```python
df['FuelType'].value_counts()
```

```python
df['FuelType'] = df['FuelType'].replace({
    'Unleaded' : 'Petrol',
    'Premium' : 'Petrol'
})
```

```python
df['FuelType'].value_counts()
```

```python
brand_fix = {
    'Land':'Land Rover',
    'Great':'Great Wall',
    'GWM':'Great Wall'
}
df['Brand'] = df['Brand'].replace(brand_fix)
```

```python
df['Brand'].unique()
```

```python
df = df.rename(columns={"Kilometres":"Kilometers"})
```

```python
df.to_csv(
    "Cleaned dataset.csv",
    index = False
)
```
