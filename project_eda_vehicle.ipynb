```python
import pandas as pd
import plotly.express as px
```

```python
df=pd.read_csv("Cleaned dataset.csv")
```

```python
count = df['Car_Condition'].value_counts().reset_index()
count
```

```python
fig = px.bar(count,x = "Car_Condition",y = "count")
```

```python
fig.show()
```

```python
fig = px.histogram(
    df,
    x="Price",
    nbins=12,
    title="Vehicle Listing Price Distribution",
    template='plotly_white',
    color_discrete_sequence=px.colors.qualitative.Set2

)
fig.update_traces(
    marker=dict(
        line=dict(color="black",width=1)
    ),
#     hovertemplate=
#     "<b>Price:</b> %{x:,.0f}<br>"
#     "<b>VehicleAge:</b> %{y}<extra></extra>"
)



fig.update_layout(
    bargap = 0.2,
    title=dict(
        
        text="Vehicle Age vs. Selling Price Progression",
        x=0.5,
        font=dict(size=24)
))
fig.show()
```

```python
fig=px.scatter(
    df,
    x="Kilometers",
    y="Price",
    color="Car_Condition",
    title="Kilometers vs Selling Price",
    template="plotly_white",
    
    )
fig.update_traces(
   marker=dict(
    symbol='triangle-up',
    size=7
   ) 
)
fig.update_layout(
    title=dict(
        text="Vehicle Age vs. Selling Price Progression",
        x=0.5,
        font=dict(size=24)
    ))
```

```python
df_brand = (
    df.groupby("Brand")["Price"].mean().sort_values(ascending=False).head(15).reset_index()
)
```

```python
fig=px.bar(
    df_brand,
    x = "Price",
    y = "Brand",
    orientation='h',
    title="Top 15 Brands by Average Market Price",
    labels={"Price":"Average Price"},
    color='Price',
    color_continuous_scale="Viridis"
    )
fig.update_layout(
    title=dict(
        text="Vehicle Age vs. Selling Price Progression",
        x=0.5,
        font=dict(size=24)
    ))
fig.update_yaxes(categoryorder="total ascending")
fig.show()
```

```python
df_age = df.groupby("VehicleAge")["Price"].mean().reset_index()

fig4 = px.line(
    df_age,
    x = "VehicleAge",
    y = "Price",
    title = "Vehicle Age vs. Selling Price Progression",
    template="plotly_dark",

    labels ={
        "VehicleAge":"Vehicle Age (Years)",
        "Price":"Average Price"
    }
)
    
fig4.update_layout(
    title=dict(
        text="Vehicle Age vs. Selling Price Progression",
        x=0.5,
        font=dict(size=24)
    ))
fig4.show()
```

```python
fig5 = px.box(
    df,
    x="Car_Condition",
    y="Price",
    color="Car_Condition",
    title="Price Distribution Matrix by Condition Status"
)
fig5.update_layout(
    title=dict(
        text="Price Distribution Matrix by Condition Status",
        x=0.5,
        font=dict(size=24)
    ))
fig5.show()
```

```python
top5_brands = df['Brand'].value_counts().head().index
df = df[df['Brand'].isin(top5_brands)]


fig = px.violin(
    df,
    x = "Brand",
    y = "VehicleAge",
    color = "Brand",
    box = True,
    points = "outliers",
    title = " Age Profile Variances across Vehicle Brands",
    color_discrete_sequence=px.colors.qualitative.Dark24
)
fig.update_layout(
    title=dict(
        text="Age Profile Variances across Vehicle Brands",
        x=0.5,
        font=dict(size=24)
    ),
yaxis_title="Vehicle Age",
width=700,
height=500,

margin=dict(
    l=70,
    r=40,
    t=80,
    b=70
)

)
fig.show()
```

```python
fig = px.scatter(
    df,
    x = "EngineSizeL",
    y = "FuelConsumption_L100km",
    color = "Cylinders",
    title = "Engine Size vs Fuel Consuption"
)
fig.update_layout(
    xaxis_title="Engine size (L)",
    yaxis_title="Fuel Consuption (L/100km)",
    template="plotly_white",
    height=600,
    width=900,

    title=dict(
        text="Engine Size vs Fuel Consuption",
        x=0.5,
        font=dict(size=24)
    ))

fig.show()
```

```python
fuel_count = (
    df["FuelType"].value_counts().reset_index()
)

fuel_count.columns = ['FuelType','Count']

#Donut chart
fig = px.pie(
    fuel_count,
    names = "FuelType",
    values = "Count",
    hole = 0.4,
    title = "Fuel Type Market Share",
    color_discrete_sequence = px.colors.qualitative.Dark24
)

fig.update_traces(
    textposition="inside",
    textinfo='percent+label',
    
    hovertemplate="<b>%{label}</b><br>count: %{value}<br>share: %{percent}<extra></extra>"
    )

fig.update_layout(
    template = "plotly_white",
    legend_title = "FuelType",
    height = 600,
    width = 900,
    
    title=dict(
        text="Fuel Type Market Share",
        x=0.5,
        font=dict(size=24)
    )
    )

fig.show()
```

```python
fig = px.box(
    df,
    x = "Transmission",
    y = "FuelConsumption_L100km",
    color="Transmission",
    points = "outliers",
    title = "Fuel Consumption by Transmission Type"
)
fig.update_layout(
    template = "plotly_white",
    yaxis_title="Fuel Consumption (L/100km)",
    height = 500,
    width = 800,

    title=dict(
        text="Fuel Consumption by Transmission Type",
        x=0.5,
        font=dict(size=24)
    )
)
```

```python
fig = px.histogram(
    df,
    x = "EngineSizeL",
    color="Transmission",
    barmode="overlay",
    nbins=25,
    opacity=0.75,
    marginal="box",
    title = "Engine Size Distributions by Transmission",
    template="plotly_dark",
   
    # color_discrete_sequence=px.colors.qualitative.
)
fig.update_layout(
    template = "plotly_white",
    xaxis_title="Engine Size (L)",
    yaxis_title="Number of Vehicles",
    legend_title='Transmission',
    height = 500,
    width = 800,
    bargap=0.2,

    title=dict(
        text="Engine Size Distributions by Transmission",
        x=0.5,
        font=dict(size=24)
    )

)

fig.update_traces(
    hovertemplate="<b>Engine Size:</b>%{x} L<br><b> %{y}<extra></extra>"
)
fig.show()
```

```python
body = (
    df.groupby(["State","BodyType"]).size().reset_index(name="Count")
)

fig = px.box(
    body,
    x = "State",
    y="Count",
    color="BodyType",
    title="Body Type Preferences Across State",

    color_discrete_sequence=px.colors.qualitative.Dark24
)

fig.update_layout(
    template="plotly_white",
    yaxis_title="Number of Vehicles",
    legend_title="BodyType",
    height=650,
    width=900
)
fig.show()
```
